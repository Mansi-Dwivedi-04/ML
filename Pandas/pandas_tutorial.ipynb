* Series: Stores 1D array homogeneous data. Size caanot be changed
* DataFrame: Store 2D data where columns can be of different types but the value within a column must remain same

```python
# Interpolate is used to handle null values more efficiently as compared to fillna which uses constant values
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "A": [1, np.nan, 3, 4, np.nan, 68],
    "B": [np.nan, 10, np.nan, 12, 13, np.nan]
})

print("Original DataFrame:")
print(df)

# Interpolate using the default linear method
df_interpolated = df.interpolate()
print("\nDataFrame after linear interpolation:")
print(df_interpolated)

# Interpolate using the 'nearest' method
df_nearest = df.interpolate(method='nearest')
print("\nDataFrame after nearest interpolation:")
print(df_nearest)
```
