```python
import matplotlib.pyplot as plt
import numpy as np
```

### Line plot

```python
x = [1,2,3,4]
y = [5,6,7,8]
plt.plot(x,y,color='red',linewidth=2,label='Line',linestyle=':',marker='x')
plt.title("Line Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.legend()
plt.show()
```

## Creating 2 bar charts using subplot

```python
np.random.seed(42)

fig, ax = plt.subplots(figsize = (10,5),nrows=1,ncols=2)
categories_1 = np.linspace(10,50,5)
categories_2 = np.linspace(50,100,5)
arr_1 = [np.random.randint(40,400) for i in range(5)]
arr_2 = [np.random.randint(70,640) for i in range(5)]

ax[0].bar(categories_1,arr_1,label="Count",width=2.5)
ax[0].set_title("Bar Chart 1")
ax[0].set_xlabel("Categories 1")
ax[0].set_ylabel("Count")
ax[0].legend()

ax[1].bar(categories_2,arr_2,color="red",label="Count",width=2.5)
ax[1].set_title("Bar Chart 2")
ax[1].set_xlabel("Categories 2")
ax[1].set_ylabel("Count")
ax[1].legend()

plt.tight_layout()
plt.show()
```

### Creating horizotal bar graph

```python
plt.barh(['Maths', 'Science', 'Finance'], [10, 20, 26],label = "Number")
plt.xlabel("Count of Students")
plt.ylabel("Subjects")
plt.title("Number of students having greater than 90% marks.")
plt.legend()
plt.show()
```

### Histogram

```python
data = np.random.randn(1000)
plt.hist(data,bins=30,color='green',edgecolor='black')
plt.grid(color='red',alpha=0.2)
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Histogram")
plt.show()
```

### Adding Labels and annotate data

```python
plt.plot([1, 2, 3], [4, 5, 6])
plt.text(2, 5, "Here", fontsize=12, color='green')
plt.show()
```

```python
plt.plot([1, 2, 3], [4, 5, 6])
plt.annotate("Peak", xy=(2, 5),xytext=(2,6),
             arrowprops=dict(facecolor='black', shrink=0.05))
plt.show()
```

### Scatter plot

```python
x = [i for i in range(1,6)]
y = [i for i in range(10,20,2)]
size = [4**i for i in range(2,7)]
color = ['blue','green','red','violet','pink']
plt.grid(alpha=0.5)
plt.scatter(x,y,s=size,alpha = 0.6,c=color)
plt.title("Scatter plot")
plt.show()
```

### Line Plot with multiple lines

```python
x = np.linspace(50,100,5)
y1 = np.linspace(40,60,5)
y2 = np.linspace(50,70,5)
plt.plot(x,y1,color='red',label='line 1')
plt.plot(x,y2,color='green',label='line 2')
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.legend()
plt.title("Multiple lines on same figure")
plt.show()
```

### Customized legend and customize a style using plt.style.use

```python
x = [1,2,3,4,5]
y = [10,15,20,25,30]
plt.style.use('ggplot')
plt.scatter(x,y,label="weekly data",s=100)
plt.legend(title='Finance',loc='upper left',fontsize=10)
plt.show()
```

### Errorbar

```python
x = [1,2,3]
y = [4,5,6]
error=[0.2,0.3,0.4]
plt.errorbar(x,y,xerr=error)
plt.show()
```

### Using twinx

```python
fig, ax = plt.subplots()

x = [1,2,3,4,5]
y1 = [1,2,3,4,5]
y2 = [100,200,300,400,500]

ax1 = ax.twinx()
ax.plot(x,y1)
ax1.plot(x,y2)
ax.set_ylabel("Low Scale")
ax1.set_ylabel("High Scale")
plt.title("Dual Axis Plotting")
ax.set_xlabel("X- Axis")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.dates as mdates

# Sample time series data
data = {'Date': pd.date_range(start='2025-01-01', periods=5),
        'Value': [10, 20, 15, 30, 25]}
df = pd.DataFrame(data)

# Plot
fig, ax = plt.subplots()
ax.plot(df['Date'], df['Value'], color='tomato')

# Date formatting
ax.xaxis.set_major_formatter(mdates.DateFormatter('%b %d'))  # e.g., Jan 01
# You can also try '%Y-%m-%d' or '%d %b' or any format

plt.xticks(rotation=45)
plt.xlabel('Date')
plt.ylabel('Value')
plt.title('Time Series Plot')
plt.tight_layout()
plt.show()
```

### Colormap and Heatmap

```python
x = np.random.rand(50)
y = np.random.rand(50)
colors=np.random.rand(50)
plt.scatter(x,y,c=colors,cmap='inferno')
plt.colorbar()
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()
```
